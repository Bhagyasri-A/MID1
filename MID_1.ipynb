```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
subjects = ["AI", "CN", "COA", "EDA", "OE"]
planned_marks = [90,75,77,75,82]
actual_marks =[82,90,85,85,75]
planned_marks.append(planned_marks[0])
actual_marks.append(actual_marks[0])
angle = np.linspace(0,2*np.pi,len(actual_marks))
plt.subplot(polar = True)
plt.thetagrids(range(0,360,int(360/len(subjects))),subjects)
plt.plot(angle, planned_marks, label = "planned marks")
plt.fill(angle, planned_marks, color = "red")
plt.plot(angle, actual_marks, label = "actual marks")
plt.fill(angle, actual_marks, color = "black")
plt.legend()
```
